## **Investigacion sobre criptografía por el método ElGamal**

La criptografía por el método ElGamal es un sistema de cifrado asimétrico (usa clave pública y clave privada) que se basa en problemas matemáticos difíciles de resolver.

Fue creado por el criptógrafo Taher ElGamal en 1985.

## **Idea principal**

El método ElGamal permite:

- Cifrar mensajes
- Firmar digitalmente
- Compartir información de forma segura

Funciona sobre la dificultad de resolver el logaritmo discreto, que es computacionalmente muy difícil cuando se usan números grandes.

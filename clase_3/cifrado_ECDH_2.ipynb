```python
from cryptography.hazmat.primitives.asymmetric.x25519 import X25519PrivateKey
from cryptography.hazmat.primitives.asymmetric import x25519
from cryptography.hazmat.primitives import serialization
import base64
from cryptography.hazmat.primitives.kdf.hkdf import HKDF
from cryptography.hazmat.primitives import hashes
import os
# Generate a private key for Alice
alice_private_key = X25519PrivateKey.generate()
alice_public_key = alice_private_key.public_key()
print("\nSe imprime la clave publica en formato RAW (crudo)\n")
raw_pub_alice = alice_public_key.public_bytes(
encoding=serialization.Encoding.Raw,
format=serialization.PublicFormat.Raw
)
print(raw_pub_alice.hex())
# Generate a private key for Bob
bob_private_key = X25519PrivateKey.generate()
bob_public_key = bob_private_key.public_key()
print("\nSe imprime la clave publica en formato PEM (mas amigable)")
pem_pub_bob = bob_public_key.public_bytes(
encoding=serialization.Encoding.PEM,
format=serialization.PublicFormat.SubjectPublicKeyInfo
)
print(pem_pub_bob.decode())
# Alice calculates the shared secret using her private key and Bob's public key
shared_secret_alice = alice_private_key.exchange(bob_public_key)
# Bob calculates the shared secret using his private key and Alice's public key
shared_secret_bob = bob_private_key.exchange(alice_public_key)
# The shared secrets should be the same
assert shared_secret_alice == shared_secret_bob
# It is best practice to use a Key Derivation Function (KDF) on the shared secret
# before using it as an encryption key.
derived_key = HKDF(
algorithm=hashes.SHA256(),
length=32,
salt=None,
info=b'handshake data',
).derive(shared_secret_alice)
# Use 'derived_key' for symmetric encryption (e.g., AES-GCM or XChaCha20-Poly1305)
print(type(shared_secret_alice))
print(len(shared_secret_alice))
print(type(bob_public_key))
print(derived_key)
```

```python
'''Funcion que verifica si P es primo
def primo(candidato_primo):
    divisores = [1]
    for i in range(2,candidato_primo):
        if candidato_primo % i == 0:
            divisores.append(i)

    if len(divisores)>2:
        return False
    else:
        return True'''

#Hay otro metodo mas eficiente para verificar si un numero es primo
# Itera menos veces, al hacer el cifrado criptografico generalmente P es un numero muy grande
# Por eso debemos ver la forma de reducir la cantidad de iteraciones

import math
def es_primo(n):
    if n<=1: # el uno no es primo
        return False
    for i in range(2, int(math.sqrt(n))+1):
        if n % i == 0:
            return False
    return True
```

```python
#Verificacion de curva eliptica

def curva_eliptica(a, b, p):
    # P tiene que ser primo y distinto de 2 y 3
    if not es_primo(p) or p in [2,3]:
        return "Error: P debe ser primo distinto de 2 y 3"
    
    # Verificar deiscriminante distinto de cero (mod p)
    delta = (4*(a**3) + 27*(b**2)) % p

    if delta == 0 :
        return "Error: La curva es singular (discriminante = 0 (mod p))"

    # ahora debemos recorrer todos los valores que puede tener Fp para calcular los puntos racionales
    # solo calcularemos el lado derecho de la ecuacion 

    #Iteracoin para calcular los puntos racionale
    puntos = []

    for x in range(p):
        lado_derecho = (x**3 + a*x + b) % p 

        for y in range(p):
            if (y**2) % p == lado_derecho:
                puntos.append((x,y))

    # Recordar el punto en el infinito
    puntos.append("0")

    return {
        " E(Fp) ": puntos,
        " #E(Fp) ": len(puntos)
    }
```

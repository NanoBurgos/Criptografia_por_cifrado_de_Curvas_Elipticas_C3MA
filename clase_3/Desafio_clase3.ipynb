```python
# Funcion que verifica si P es primo
def primo(candidato_primo):
    divisores = []
    for i in range(2,candidato_primo,1):
        if candidato_primo % i == 1:
            divisores.append(i)

    if len(divisores)>1:
        return False
    else:
        return True
```
